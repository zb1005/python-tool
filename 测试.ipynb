```python
import pandas as pd
df = pd.read_excel(r'C:\Users\zhangbon\Desktop\1、PCT-物料渠道信息.xlsx')
```

```python
df = df.fillna('')
# 渠道状态拆分结果中，如果最后一个元素为空，删除
# 定义处理函数：拆分后若最后一个元素为空则删除
def clean_last_empty(lst):
    # 先过滤掉拆分后可能的空字符串（如连续分号的情况），再判断最后一个元素
    # 或仅判断最后一个元素是否为空：
    if lst and lst[-1] == "":  # 列表非空且最后一个元素为空
        return lst[:-1]
    return lst
df['渠道状态拆分结果'] = df['渠道状态'].str.split(';').apply(clean_last_empty)
df
df_exploded = df.explode('渠道状态拆分结果').reset_index(drop=True)
df_exploded
```

```python
df_exploded = df_exploded.fillna('')
df_exploded['渠道'] = df_exploded['渠道状态拆分结果'].str.split(',').str[0].fillna('')
df_exploded['状态'] = df_exploded['渠道状态拆分结果'].str.split(',').str[1].fillna('')
df_exploded
```
