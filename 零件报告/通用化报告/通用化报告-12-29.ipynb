### 总的零件通用化

#### 筛选生产过的整机

```python
# 筛选出有过生产的整机相关数据
worked_set = set(df_work_list['物料编码'])
df_mbom_worked = df_merged_mbom_plm[df_merged_mbom_plm['最终父项物料编码'].isin(worked_set)].reset_index(drop=True)
# 保留最终父和子，并去重
df_mbom_worked = df_mbom_worked[['最终父项物料编码','子项物料编码','标准型号','产品组','国内/海外']].drop_duplicates().reset_index(drop=True)
```

#### 整机的筛选（标准型号）

```python
# 每个标准型号只保留最老的产品数据
stand_old_data  = df_mbom_worked.groupby('标准型号',as_index=False).agg(
                                                        最终父项物料编码  = ('最终父项物料编码', 'min')            
)
stand_old_data

# 对于每个标准型号只保留最老的产品数据
df_calu1 = df_mbom_worked[df_mbom_worked['最终父项物料编码'].isin(stand_old_data['最终父项物料编码'])].reset_index(drop=True)
df_calu1['产品组'].value_counts()
df_calu1
print(len(df_calu1))
# 剔除海外的物料
df_calu1 = df_calu1[df_calu1['国内/海外']=='国内']
df_calu1
print(len(df_calu1))
```

```python
df_calu1['产品集合'] = df_calu1['产品组'].map(productgroupsset_map_1)
df_calu1
```

```python
df_calu1.to_excel(fr"D:\000物料报表\通用化报告\{month}\明细表-整机和零件（整机和零件都进行了筛选）.xlsx",index=False)
```

```python
# 统计每个产品组内部，各个零部件的共同型号数
df_result1 = df_calu1.groupby(['产品集合','子项物料编码'],as_index=False).agg(
                                                        零件共用型号数=('最终父项物料编码','nunique'),
                                                        零件归属型号=('最终父项物料编码',lambda x: list(x.unique()))
)
df_result1.to_excel(fr"D:\000物料报表\通用化报告\{month}\零部件共用型号统计.xlsx",index=False)
df_result1
```

```python
df_result2 = df_result1.groupby(['产品集合'],as_index=False).agg(
                                                        通用化零件数=('零件共用型号数',lambda x: x[x>1].count()),
                                                        通用2次以上的零件数=('零件共用型号数',lambda x: x[x>=2].count()),
                                                        通用4次以上的零件数=('零件共用型号数',lambda x: x[x>=4].count()),
                                                        通用6次以上的零件数=('零件共用型号数',lambda x: x[x>=6].count()),
                                                        通用8次以上的零件数=('零件共用型号数',lambda x: x[x>=8].count()),
                                                        零件涉及产品型号数 = ('零件归属型号',lambda x: len(set().union(*x))),
                                                        零部件数=('子项物料编码','nunique')
)
df_result2['通用化零件数占比'] = df_result2['通用化零件数']/df_result2['零部件数']
productset_sort = {'吸油烟机':0,
                    '灶具':1,
                    '蒸烤微合计':2,
                    '灶集成':3,
                    '消毒柜':4,
                    '热水器':5,
                    '水槽洗碗机':6,
                    '嵌入式洗碗机':7,
                    '家用净水机':8}
df_result2.sort_values(by='产品集合',key=lambda x: x.map(productset_sort),inplace=True)
df_result2 = df_result2.reset_index(drop=True)
df_result2
```

```python
def mlutone(li):
    if set(li) == {'电机', '电机支架'} :
        return '电机支架'
    elif set(li) == {'出风罩', '密封垫'} :
        return '密封垫'
    elif set(li) == {'蜗壳', '吸音棉'} :
        return '吸音棉'
    elif set(li) == {'蜗壳', '吸音盒'}  :
        return '吸音盒'
    elif set(li) == {'蜗壳', '密封垫'} :
        return '密封垫'
    elif set(li) == {'吸音棉', '风机盖'} :
        return '吸音棉'
    else:
        return li[0] if len(li) == 1 else li
```

```python
df_calu2['管控零件族简称'] = df_calu2['管控零件族简称'].apply(lambda x: mlutone(x))
df_calu2
```

```python
df_result3 = df_calu2.groupby(['管控零件族简称','子项物料编码'],as_index=False).agg(
                                                                零件共用型号数=('最终父项物料编码', 'nunique'),
                                                                零件归属型号 = ('最终父项物料编码', lambda x: list(x.unique())),
                                                                零件共用物料组数 = ('物料组描述', 'nunique'),
                                                                零件共用物料组 = ('物料组描述', lambda x: list(x.unique()))
)
df_result3
```

```python
df_result4 = df_result3.groupby('管控零件族简称',as_index=False).agg(
                                                        通用化零件数=('零件共用型号数',lambda x: x[x>1].count()),
                                                        通用2次以上的零件数=('零件共用型号数',lambda x: x[x>=2].count()),
                                                        通用3次以上的零件数=('零件共用型号数',lambda x: x[x>=3].count()),
                                                        通用4次以上的零件数=('零件共用型号数',lambda x: x[x>=4].count()),
                                                        通用5次以上的零件数=('零件共用型号数',lambda x: x[x>=5].count()),
                                                        通用6次以上的零件数=('零件共用型号数',lambda x: x[x>=6].count()),
                                                        零部件数=('子项物料编码','nunique'),
                                                        涉及的型号数=('零件归属型号',lambda x: len(set().union(*x))),
).reset_index(drop=True)
df_result4.to_excel(fr"D:\000物料报表\通用化报告\{month}\通用化统计报告-零件族-油烟机-产品型号.xlsx",index=False)
df_result4
```
