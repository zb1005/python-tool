### 数据清洗

```python
df_mbom = pd.read_excel(fr"D:\000物料报表\通用化报告\1003工厂mbom多行展开.XLSX")
df_work_list = pd.read_excel(fr"D:\000物料报表\通用化报告\烹饪-1003-251031-生产工单.XLSX")
```

```python
# 处理数据格式
df_mbom['最终父项物料编码'] = df_mbom['最终父项物料编码'].astype(str).str[:13]
df_work_list['物料编码'] = df_work_list['物料编号'].astype(str).str[:13]
```

```python
# 筛选出有过生产的整机相关数据
worked_set = set(df_work_list['物料编码'])
df_mbom_worked = df_mbom[df_mbom['最终父项物料编码'].isin(worked_set)].reset_index(drop=True)
# 保留最终父和子，并去重
df_mbom_worked = df_mbom_worked[['最终父项物料编码','子项物料编码']].drop_duplicates().reset_index(drop=True)
```

```python
df_mbom_worked
```

```python
df_plm = pd.read_excel(fr"D:\000物料报表\通用化报告\产品生命周期状态全表.xlsx")
df_plm['物料号'] = df_plm['物料号'].astype(str).str.strip().str[:13]
df_plm
```

```python
# 把产品相关信息匹配进去
df_calu = pd.merge(df_mbom_worked, df_plm[['物料号', '标准型号', '国内/海外', '产品组']].drop_duplicates().reset_index(drop=True), left_on='最终父项物料编码',right_on='物料号', how='left')
df_calu[df_calu['最终父项物料编码'].isnull()]
print(len(df_calu))
```

```python
# 每个标准型号只保留最老的产品数据
stand_old_data  = df_calu.groupby('标准型号',as_index=False).agg(
                                                        最终父项物料编码  = ('最终父项物料编码', 'min')            
)
stand_old_data

# 对于每个标准型号只保留最老的产品数据
df_calu1 = df_calu[df_calu['最终父项物料编码'].isin(stand_old_data['最终父项物料编码'])].reset_index(drop=True)
df_calu1['产品组'].value_counts()
df_calu1
print(len(df_calu1))
# 剔除海外的物料
df_calu1 = df_calu1[df_calu1['国内/海外']=='国内']
df_calu1
print(len(df_calu1))
```

```python
df_calu1.to_excel(fr'D:\000物料报表\通用化报告\需要导出信息的零件集合.xlsx',index=False)
```

```python
# 剔除掉虚拟物料和服务专用配件
df_filter1 = pd.read_excel(fr"D:\000物料报表\通用化报告\虚拟-服务专用配件-图号判断.XLSX")
df_calu1['物料描述'] = df_calu1['子项物料编码'].map(dict(zip(df_filter1['物料编码'], df_filter1['物料描述'])))
df_calu1['图号'] = df_calu1['子项物料编码'].map(dict(zip(df_filter1['物料编码'], df_filter1['图号'])))
df_calu1['是否服务专用配件'] = df_calu1['子项物料编码'].map(dict(zip(df_filter1['物料编码'], df_filter1['是否服务专用配件'])))
df_calu1['是否虚拟物料'] = df_calu1['子项物料编码'].map(dict(zip(df_filter1['物料编码'], df_filter1['是否虚拟物料'])))
print(len(df_calu1))
# 剔除掉虚拟物料和服务专用配件
df_calu1 = df_calu1[(df_calu1['是否服务专用配件']!='X') & (df_calu1['是否虚拟物料']!='X')].reset_index(drop=True)
print(len(df_calu1))
# 只保留11开头的子项物料编码
df_calu1 = df_calu1[df_calu1['子项物料编码'].str.startswith('11')].reset_index(drop=True)
print(len(df_calu1))
# 剔除图号结尾是-0001，-0002，-0003...-0009,到-0099的物料
df_calu1 = df_calu1[~df_calu1['图号'].str.contains(r'-00\d{2}$', na=False)]
print(len(df_calu1))
```

```python
# 剔除工艺中间件、散装物料、关键字剔除
df_filter2 = pd.read_excel(fr"D:\000物料报表\通用化报告\物料_零件族.xlsx")
df_filter2['物料编码'] = df_filter2['物料编码'].astype(str).str.strip().str[:13]
df_calu1['零件族'] = df_calu1['子项物料编码'].map(dict(zip(df_filter2['物料编码'], df_filter2['零件族'])))
df_calu1
print(len(df_calu1))
# 剔除零件族种包含标贴、说明书、信息码、接线图、附件、插页、铭牌、包装图、公用、标识、共用、手册、操作指引、操作说明、菜谱、印刷品
df_calu1 = df_calu1[~df_calu1['零件族'].str.contains('印刷品|壳体组件-消毒柜|钣金组件-消毒柜', na=False)].reset_index(drop=True)
df_calu1
print(len(df_calu1))
```

```python
# 聚合一个新的产品组，作为统计维度
df_calu1['产品组1'] = df_calu1['产品组'].map(productgroupsset_map_1)
df_calu1['产品组1'].value_counts(),len(df_calu1)
df_calu1.to_excel(fr"D:\000物料报表\通用化报告\明细表.xlsx",index=False)
```

```python
# 统计每个产品组内部，各个零部件的共同型号数
df_result1 = df_calu1.groupby(['产品组1','子项物料编码'],as_index=False).agg(
                                                        零件共同型号数=('最终父项物料编码','nunique'),
                                                        零件归属型号数=('最终父项物料编码',lambda x: list(x.unique()))
)
df_result1
```

```python
df_result2 = df_result1.groupby(['产品组1'],as_index=False).agg(
                                                        通用化零件数=('零件共同型号数',lambda x: x[x>1].count()),
                                                        通用2次以上的零件数=('零件共同型号数',lambda x: x[x>=2].count()),
                                                        通用3次以上的零件数=('零件共同型号数',lambda x: x[x>=3].count()),
                                                        通用4次以上的零件数=('零件共同型号数',lambda x: x[x>=4].count()),
                                                        通用5次以上的零件数=('零件共同型号数',lambda x: x[x>=5].count()),
                                                        通用6次以上的零件数=('零件共同型号数',lambda x: x[x>=6].count()),
                                                        零部件数=('子项物料编码','nunique')

)
df_result2
```

```python
df_result2['通用化零件数占比'] = df_result2['通用化零件数']/df_result2['零部件数']
df_result2.to_excel(fr"D:\000物料报表\通用化报告\通用化统计报告.xlsx",index=False)
```
